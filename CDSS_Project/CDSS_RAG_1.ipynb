Find all pdf files and Extract text from each PDF

```python
from pathlib import Path
from pypdf import PdfReader

project_dir = Path.cwd()

documents_dir = project_dir / "RAG" / "documents"

if not documents_dir.exists():
    raise FileNotFoundError(
        f"Documents folder not found: {documents_dir}\n"
        "Make sure the notebook is running from your project directory."
    )

pdf_files = sorted(documents_dir.glob("*.pdf"))

print("Documents folder:", documents_dir)
print("Number of PDF files found:", len(pdf_files))

if not pdf_files:
    print("No PDF files found. Check the folder and file extensions.")

documents = []

for pdf_path in pdf_files:
    try:
        reader = PdfReader(str(pdf_path))
        page_texts = []

        for page_number, page in enumerate(reader.pages, start=1):
            text = page.extract_text() or ""

            if text.strip():
                page_texts.append(
                    f"[Page {page_number}]\n{text.strip()}"
                )

        full_text = "\n\n".join(page_texts)

        documents.append({
            "source": pdf_path.name,
            "text": full_text,
            "pages": len(reader.pages)
        })

        print(
            f"\nFile: {pdf_path.name}"
            f"\nPages: {len(reader.pages)}"
            f"\nCharacters extracted: {len(full_text):,}"
        )

        if not full_text.strip():
            print("WARNING: No text extracted. This PDF may be scanned.")

    except Exception as error:
        print(f"\nCould not read {pdf_path.name}: {error}")

print("\nTotal documents processed:", len(documents))
print(
    "Total extracted characters:",
    f"{sum(len(doc['text']) for doc in documents):,}"
)
```

Split extracted PDF text into chunks

```python
CHUNK_SIZE = 1000
CHUNK_OVERLAP = 150

chunks = []

for document in documents:
    text = document["text"]
    source = document["source"]

    pages = text.split("[Page ")

    for page_section in pages:
        if not page_section.strip():
            continue

        page_number_text, separator, page_text = (
            page_section.partition("]")
        )

        if not separator:
            continue

        try:
            page_number = int(page_number_text.strip())
        except ValueError:
            continue

        page_text = page_text.strip()

        if not page_text:
            continue

        start = 0

        while start < len(page_text):
            end = min(start + CHUNK_SIZE, len(page_text))
            chunk_text = page_text[start:end].strip()

            if chunk_text:
                chunks.append({
                    "text": chunk_text,
                    "source": source,
                    "page": page_number
                })

            if end >= len(page_text):
                break

            start = end - CHUNK_OVERLAP

print("Total chunks created:", len(chunks))

if chunks:
    print("\nFirst chunk source:", chunks[0]["source"])
    print("First chunk page:", chunks[0]["page"])
    print("\nFirst chunk preview:\n")
    print(chunks[0]["text"][:500])
```

Connect to pinecone

```python
import os
from dotenv import load_dotenv
from pinecone import Pinecone
from pathlib import Path

project_dir = Path.cwd()
load_dotenv(project_dir / ".env")

api_key = os.getenv("PINECONE_API_KEY")

if not api_key:
    raise ValueError(
        "Pinecone API key not found. Check your .env file."
    )

pc = Pinecone(api_key=api_key)

print("Successfully connected to Pinecone.")
```

```python
from pinecone import Pinecone, ServerlessSpec

index_name = "heart-disease-knowledge"

if not pc.indexes.exists(index_name):
    pc.create_index(
        name=index_name,
        dimension=384,
        metric="cosine",
        spec=ServerlessSpec(
            cloud="aws",
            region="us-east-1"
        )
    )
    print("Pinecone index created.")
else:
    print("Pinecone index already exists.")

index = pc.Index(index_name)

print("Connected to:", index_name)
```

Load a lightweight embedding model

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2",
    device="cpu"
)

print("Embedding model loaded successfully.")
```

Test the embedding model with a sample sentence

```python
sample_text = (
    "High blood pressure is a risk factor for heart disease."
)

sample_embedding = embedding_model.encode(sample_text)

print("Embedding generated successfully.")
print("Vector dimensions:", sample_embedding.shape)
print("First 10 values:", sample_embedding[:10])
```

```python
import hashlib
import numpy as np

if not chunks:
    raise ValueError("No text chunks found. Run the chunking cell first.")

print("Total chunks to upload:", len(chunks))

batch_size = 32
total_uploaded = 0

for start in range(0, len(chunks), batch_size):

    batch = chunks[start:start + batch_size]

    texts = [item["text"] for item in batch]

    embeddings = embedding_model.encode(
        texts,
        batch_size=16,
        convert_to_numpy=True,
        normalize_embeddings=True,
        show_progress_bar=False
    )

    vectors = []

    for item, embedding in zip(batch, embeddings):

        unique_text = (
            f'{item["source"]}|'
            f'{item["page"]}|'
            f'{item["text"]}'
        )

        vector_id = hashlib.sha256(
            unique_text.encode("utf-8")
        ).hexdigest()

        vectors.append({
            "id": vector_id,
            "values": embedding.tolist(),
            "metadata": {
                "text": item["text"],
                "source": item["source"],
                "page": int(item["page"])
            }
        })

    # Upload this batch to Pinecone
    index.upsert(vectors=vectors)

    total_uploaded += len(vectors)

    if total_uploaded % 320 == 0 or total_uploaded == len(chunks):
        print(f"Uploaded {total_uploaded}/{len(chunks)} chunks")

print("\nEmbedding generation and upload completed.")
```

Check the current Pinecone index statistics

```python

stats = index.describe_index_stats()

print("Total vectors stored:", stats.total_vector_count)
print("Index dimension:", stats.dimension)
```

Medical question to test retrieval

```python
query = "What are the common risk factors for heart disease?"

# Converting the question into an embedding
query_embedding = embedding_model.encode(
    query,
    convert_to_numpy=True,
    normalize_embeddings=True
).tolist()

# Search Pinecone for the 5 most relevant chunks
search_results = index.query(
    vector=query_embedding,
    top_k=5,
    include_metadata=True
)

# Display the retrieved passages
print("Question:", query)
print("\nRetrieved medical evidence:\n")

for rank, match in enumerate(search_results["matches"], start=1):
    metadata = match["metadata"]

    print(f"Result {rank}")
    print(f"Similarity score: {match['score']:.4f}")
    print(f"Source PDF: {metadata['source']}")
    print(f"Page: {metadata['page']}")
    print(f"Text: {metadata['text'][:700]}")
    print("-" * 80)
```
